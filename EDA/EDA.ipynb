```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("../Data/CreditWorthiness.xlsx")
```

```python
df.sample(2)
```

```python
#important features
df = df[['Cbal', 'Cdur', 'Chist', 'Camt', 'Edur', 'InRate', 'MSG', 'Rdur',
       'Prop', 'age', 'inPlans', 'Htype', 'NumCred', 'JobType', 'Ndepend',
       'telephone', 'foreign']]
```

```python
#Performing Basic EDA
from pandas_profiling import ProfileReport
profile = ProfileReport(df)
profile.to_file(output_file="index.html")
```

```python
numerical_cols = df.describe().columns
```

```python
for col in numerical_cols:
    plt.figure(figsize=(5,4))
    sns.boxplot(df[col])
    plt.show()
```

```python
sns.pairplot(df)
```

```python
for col in numerical_cols:
    plt.figure(figsize=(5,4))
    sns.kdeplot(df[col])
    plt.title(col)
    plt.show()
```

```python
for col in numerical_cols:
    plt.figure(figsize=(5,4))
    sns.distplot(df[col])
    plt.title(col)
    plt.show()
```

```python
for col in numerical_cols:
    plt.figure(figsize=(5,4))
    plt.hist(df[col],bins=10,density=True)
    plt.title(col)
    plt.show()
```

```python
sns.heatmap(df[numerical_cols].corr())
```
